# 04 Feature Analysis

Purpose: Validate the statistical distributions and quality of engineered features.

## 1. Dataset Shape
The dataset contains 307,511 unique application rows. Overall default rate is 8.07%.

## 2. Feature Anomalies
- `APP_DAYS_EMPLOYED` contains the known Home Credit anomaly `365243` for pensioners. This will be replaced with NaN during pipeline preprocessing.
- No constant features or infinite values detected.

## 3. Thin-File Analysis
- 44,020 borrowers have 0 traditional credit records.
- Default rate for thin-files is **10.12%**, compared to 7.73% for non-thin-files.
- Crucially, **41,550** of these thin-file borrowers DO have alternative internal records, enabling our hypothesis test.

## 4. Correlations
The most highly correlated features ($r \approx 0.90$) are `INST_LATE_PAYMENT_COUNT` and `INST_UNDERPAYMENT_COUNT`. Most other feature groups are largely orthogonal.

For complete statistics, see `docs/feature_validation.md`.
