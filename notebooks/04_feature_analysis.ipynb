# 04 Feature Analysis

Purpose: Validate the statistical distributions and quality of engineered features.

## 1. Dataset Shape
The dataset contains 307,511 unique application rows. Overall default rate is 8.07%.

## 2. Feature Anomalies
- `APP_DAYS_EMPLOYED` contains the known Home Credit anomaly `365243` for pensioners. This will be replaced with NaN during pipeline preprocessing.
- No constant features or infinite values detected.

## 3. Thin-File Analysis
- 44,020 borrowers have 0 traditional credit records.
- Default rate for thin-files is **10.12%**, compared to 7.73% for non-thin-files.
- Crucially, **41,550** of these thin-file borrowers DO have alternative internal records, enabling our hypothesis test.
- *Note: Coverage does not automatically establish predictive improvement; the upcoming ablation study will test that directly.*

## 4. Correlations & Zero-Imputation Ambiguity
- The most highly correlated features ($r \approx 0.90$) are `INST_LATE_PAYMENT_COUNT` and `INST_UNDERPAYMENT_COUNT`. We verified this is partly structural since both are correlated (~0.5) with `INST_TOTAL_COUNT`. Using ratios like `INST_LATE_PAYMENT_RATIO` isolates the behavior from the volume.
- Zero-imputation creates ambiguity: `0` can mean 'clean history' or 'no history'. For example, 136,644 borrowers have 0 late payments despite having a history, while 15,868 have 0 simply because they have no history.

## 5. Missingness as a Signal
Missingness in `EXT_SOURCE_3` is associated with a 9.31% default rate vs 7.77% when present. Missingness itself is predictive.
