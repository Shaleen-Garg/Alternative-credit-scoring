# 04 Feature Analysis

Purpose: Analyze engineered credit risk features and validate the dataset construction.

## Engineered Features Overview
The feature pipeline extracted 23 strong, interpretable features focusing on:
- **Application data**: Demographics, stated income, requested amounts.
- **Traditional Credit**: Bureau trade lines, active debts, utilization.
- **Alternative Behavioural**: Internal prior loan interactions, installment promptness, and previous DPDs.

## Validation Highlights
- Total rows matched the `application_train` set precisely (307,511 rows).
- 1-to-1 mapping via `SK_ID_CURR` was maintained (no duplication).
- Temporal logic was enforced strictly up to the point of application.
- **Thin-File Segmentation**: The 44,020 borrowers lacking `bureau` records were perfectly preserved with `BUREAU_CREDIT_COUNT = 0`. Crucially, their alternative behavioral history remains intact, which is central to our hypothesis testing.

The dataset is fully prepared for modeling. See `docs/feature_specification.md` for a complete breakdown of each feature and its expected interpretation.
