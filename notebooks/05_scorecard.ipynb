# 05 Baseline Scorecard Model

Purpose: Train the first interpretable baseline credit-risk model using ONLY the Application feature group. This serves as our comparative baseline for the upcoming ablation study.

## 1. Pipeline Execution
The full modelling pipeline is implemented in `src/models.py`. It explicitly enforces a strict 70/15/15 stratified split, preventing data leakage. Preprocessing parameters (e.g., imputation means, scalers) are strictly learned on the training set.

```python
from src.models import load_and_split_data, build_baseline_pipeline
from src.evaluation import evaluate_model

# 1. Load Data
filepath = "../data/processed/feature_master.csv"
X_train, X_val, X_test, y_train, y_val, y_test = load_and_split_data(filepath)

# 2. Build & Train Model
pipeline = build_baseline_pipeline()
pipeline.fit(X_train.drop(columns=['SK_ID_CURR']), y_train)

# 3. Predict
y_test_prob = pipeline.predict_proba(X_test.drop(columns=['SK_ID_CURR']))[:, 1]

# 4. Evaluate
metrics = evaluate_model(y_test, y_test_prob)
print(metrics)
```

## 2. Test Set Results
- **ROC-AUC**: 0.7294
- **PR-AUC**: 0.2085

The model successfully captures significant predictive signal without overfitting (Train ROC-AUC: 0.7268). The external scores heavily dominate the ranking.
