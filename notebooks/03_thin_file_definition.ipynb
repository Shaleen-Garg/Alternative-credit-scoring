# 03 Thin File Definition

Purpose: Define and identify thin-file borrowers.

## Candidate Analysis
| Candidate Definition | Borrowers | % of Train | Default Rate | Comp. Default Rate |
|----------------------|----------:|-----------:|-------------:|-------------------:|
| A: No Bureau Record  | 44,020 | 14.31% | 10.12% | 7.73% |

| D2: <= 2 Bureau Records | 115,727 | 37.63% | 8.93% | 7.55% |

## Conclusion
We select **Candidate A (No Bureau Record)** as our final definition.

It perfectly matches the definition of having zero traditional credit history. 
Furthermore, 41,550 of these 44,020 borrowers *do* have alternative internal records (previous_application), providing the exact alternative data needed to test our hypothesis.
