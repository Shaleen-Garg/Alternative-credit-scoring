# 07 Controlled Feature-Group Ablation

This notebook documents and launches the Phase 7A/7B/8 comparison. The source of truth for centralized features, split construction, preprocessing, metrics, and report generation is in `src/models.py`, `src/evaluation.py`, and `src/ablation.py`.

All models use the same frozen borrower IDs in `data/processed/splits/`, train-fitted preprocessing, and `LogisticRegression(C=1.0, class_weight='balanced')`. Run from the repository root or set paths below. No tuning, calibration, threshold selection, or boosting is performed. The test set was already inspected in the original benchmark and is used only for predefined final comparisons.

The three predictor sets are the ten application features; those ten plus six bureau features; and those sixteen plus eight alternative/non-bureau behavioural features. The final feature captures missing installment payment information explicitly. Thin-file remains fixed at `BUREAU_CREDIT_COUNT == 0`.
